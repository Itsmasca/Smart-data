# Introducción a los Algoritmos Genéticos
## Actividad 16 - Comparación de poblaciones (10, 30 y 50 individuos)

**Función objetivo:** $f(x, y) = \sqrt{x^2 + y^2}$

**Dominio:** $\{x, y\} \in [-100, 100]$

**Soluciones óptimas:** $(\pm100, \pm100)$ con $f = 141.4214$

Se implementa el algoritmo genético usando la librería DEAP para tres tamaños de población: **10, 30 y 50 individuos**, con **20 generaciones** cada uno.

```python
# 2. Librerías necesarias
import random
import math
import numpy as np
from deap import base, creator, tools, algorithms
import matplotlib.pyplot as plt
from matplotlib import cm
```

## Representación gráfica de la función de optimización

```python
# Definimos la función objetivo para graficar
def funcion_prueba(x):
    res = np.sqrt(x[0] ** 2 + x[1] ** 2)
    return res

# Generamos vectores que barran todo el dominio
x = np.linspace(-100, 100, 1000)
y = np.linspace(-100, 100, 1000)

# Evaluamos todos los valores de las variables
x_ax, y_ax = np.meshgrid(x, y)
vals = np.c_[x_ax.ravel(), y_ax.ravel()]
fx = np.reshape([funcion_prueba(val) for val in vals], (1000, 1000))

# Representamos los resultados obtenidos
figure_3d = plt.figure(figsize=(10, 7))
ax = figure_3d.add_subplot(111, projection="3d")
ax.plot_surface(x_ax, y_ax, fx, cmap=cm.coolwarm)
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("f(x,y)")
ax.set_title("Espacio de búsqueda: $f(x,y) = \sqrt{x^2 + y^2}$")
plt.tight_layout()
plt.show()
```

## Codificación del algoritmo genético

### 3. Definición del problema
- **Tipo de problema de optimización:** maximización
- **Tipo de objeto:** Lista
- **Toolbox:** Función para crear individuo y población

```python
# 4. Creación del problema
# weights=(1.0,) indica que es un problema de MAXIMIZACIÓN
creator.create("FitnessMax", base.Fitness, weights=(1.0,))

# 5. Creación del individuo
# El individuo es una lista con un atributo fitness
creator.create("Individual", list, fitness=creator.FitnessMax)
```

```python
# 7. Función objetivo
def funcion_objetivo(x):
    """
    Función objetivo: f(x,y) = sqrt(x^2 + y^2)
    Devuelve -1 si los valores están fuera del dominio [-100, 100]
    Siempre devuelve una tupla.
    """
    for i in range(len(x)):
        if x[i] > 100 or x[i] < -100:
            return -1,
    res = math.sqrt(x[0]**2 + x[1]**2)
    return res,
```

### Función principal del algoritmo genético

Se encapsula todo el algoritmo en una función que recibe el tamaño de población como parámetro, para poder comparar los resultados con diferentes tamaños.

```python
def ejecutar_ag(tam_poblacion, semilla=42):
    """
    Ejecuta el algoritmo genético con un tamaño de población dado.

    Parámetros:
    - tam_poblacion: número de individuos en la población
    - semilla: semilla para el generador de números pseudoaleatorios

    Retorna:
    - hof: Hall of Fame (mejor individuo encontrado)
    - logbook: registro de estadísticas por generación
    """
    random.seed(semilla)

    toolbox = base.Toolbox()

    # 6. Crear individuos aleatorios y población inicial
    # Generación de genes: valores aleatorios uniformes entre -100 y 100
    toolbox.register("attr_uniform", random.uniform, -100, 100)

    # Creación del individuo completo (lista de 2 variables: x e y)
    toolbox.register("individual", tools.initRepeat, creator.Individual, toolbox.attr_uniform, 2)

    # Creación de la población
    toolbox.register("population", tools.initRepeat, list, toolbox.individual, tam_poblacion)

    # Registro de la función objetivo (fitness)
    toolbox.register("evaluate", funcion_objetivo)

    # 8. Operadores genéticos
    # Cruza: cruce de un punto
    toolbox.register("mate", tools.cxOnePoint)
    # Mutación: mutación gaussiana (mu=0, sigma=5, probabilidad por gen=0.1)
    toolbox.register("mutate", tools.mutGaussian, mu=0, sigma=5, indpb=0.1)
    # Selección: torneo de tamaño 3
    toolbox.register("select", tools.selTournament, tournsize=3)

    # 10. Configuración del algoritmo genético
    CXPB, MUTPB, NGEN = 0.5, 0.2, 20  # Probabilidades de cruza, mutación y número de generaciones
    pop = toolbox.population()           # Generación de la población inicial
    hof = tools.HallOfFame(1)            # Almacena al mejor individuo encontrado

    # Estadísticas de población sobre el fitness
    stats = tools.Statistics(lambda ind: ind.fitness.values)
    stats.register("avg", np.mean)    # Calcular la media
    stats.register("std", np.std)     # Desviación estándar típica
    stats.register("min", np.min)     # Valor mínimo absoluto de la generación
    stats.register("max", np.max)     # Valor máximo absoluto de la generación
    logbook = tools.Logbook()

    # Ejecutar el algoritmo genético simple
    pop, logbook = algorithms.eaSimple(
        pop, toolbox,
        cxpb=CXPB, mutpb=MUTPB, ngen=NGEN,
        stats=stats, halloffame=hof, verbose=True
    )

    return hof, logbook
```

```python
# 11. Convergencia del algoritmo
def graficar_convergencia(log, tam_poblacion):
    """
    Representa la evolución del mejor individuo en cada generación.

    Muestra tres curvas:
    - Azul: valor mínimo del fitness en cada generación
    - Rojo: valor máximo del fitness en cada generación
    - Negro (discontinua): valor promedio en cada generación
    """
    gen = np.array(log.select("gen"))
    fit_mins = np.array(log.select("min"), dtype=float)
    fit_maxs = np.array(log.select("max"), dtype=float)
    fit_ave = np.array(log.select("avg"), dtype=float)

    fig, ax1 = plt.subplots(figsize=(8, 5))
    ax1.plot(gen, fit_mins, "b", label="Mín")
    ax1.plot(gen, fit_maxs, "r", label="Máx")
    ax1.plot(gen, fit_ave, "--k", label="Promedio")
    where_mask = fit_maxs >= fit_mins
    ax1.fill_between(gen, fit_mins, fit_maxs, where=where_mask, facecolor='g', alpha=0.2)
    ax1.set_xlabel("Generación")
    ax1.set_ylabel("Fitness")
    ax1.set_ylim([-10, 160])
    ax1.set_title(f"Convergencia - Población de {tam_poblacion} individuos")
    ax1.legend(["Mín", "Máx", "Promedio"], loc="lower center")
    plt.grid(True)
    plt.tight_layout()
    plt.show()
```

## Ejecución del algoritmo genético

### Población de 10 individuos

```python
# Ejecutar con población de 10 individuos
hof_10, log_10 = ejecutar_ag(10)
print(f"\nMejor fitness: {hof_10[0].fitness.values[0]:.6f}")
print(f"Mejor individuo: x = {hof_10[0][0]:.6f}, y = {hof_10[0][1]:.6f}")
```

```python
# Gráfica de convergencia - Población de 10
graficar_convergencia(log_10, 10)
```

### Población de 30 individuos

```python
# Ejecutar con población de 30 individuos
hof_30, log_30 = ejecutar_ag(30)
print(f"\nMejor fitness: {hof_30[0].fitness.values[0]:.6f}")
print(f"Mejor individuo: x = {hof_30[0][0]:.6f}, y = {hof_30[0][1]:.6f}")
```

```python
# Gráfica de convergencia - Población de 30
graficar_convergencia(log_30, 30)
```

### Población de 50 individuos

```python
# Ejecutar con población de 50 individuos
hof_50, log_50 = ejecutar_ag(50)
print(f"\nMejor fitness: {hof_50[0].fitness.values[0]:.6f}")
print(f"Mejor individuo: x = {hof_50[0][0]:.6f}, y = {hof_50[0][1]:.6f}")
```

```python
# Gráfica de convergencia - Población de 50
graficar_convergencia(log_50, 50)
```

## Comparación de convergencia

Gráfica combinada de los tres tamaños de población para facilitar la comparación visual.

```python
# Gráfica comparativa de las tres poblaciones
fig, axes = plt.subplots(1, 3, figsize=(18, 5), sharey=True)

for i, (log, tp) in enumerate(zip([log_10, log_30, log_50], [10, 30, 50])):
    gen = np.array(log.select("gen"))
    fit_mins = np.array(log.select("min"), dtype=float)
    fit_maxs = np.array(log.select("max"), dtype=float)
    fit_ave = np.array(log.select("avg"), dtype=float)

    axes[i].plot(gen, fit_mins, "b", label="Mín")
    axes[i].plot(gen, fit_maxs, "r", label="Máx")
    axes[i].plot(gen, fit_ave, "--k", label="Promedio")
    where_mask = fit_maxs >= fit_mins
    axes[i].fill_between(gen, fit_mins, fit_maxs, where=where_mask, facecolor='g', alpha=0.2)
    axes[i].set_xlabel("Generación")
    axes[i].set_title(f"Población = {tp}")
    axes[i].set_ylim([-10, 160])
    axes[i].legend(["Mín", "Máx", "Promedio"], loc="lower center")
    axes[i].grid(True)

axes[0].set_ylabel("Fitness")
fig.suptitle("Comparación de Convergencia del Algoritmo Genético", fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

## Tabla comparativa de resultados

```python
# 14. Tabla comparativa de resultados
optimo = math.sqrt(100**2 + 100**2)  # Valor óptimo teórico: 141.421356

print(f"Valor óptimo teórico: f(±100, ±100) = {optimo:.6f}")
print()
print(f"{'Población':<12} {'Mejor Fitness':<16} {'Mejor x':<14} {'Mejor y':<14} {'Prom. Final':<13} {'Desv. Est.':<12} {'% Error':<10}")
print(f"{'-'*90}")

for tp, hof, log in [(10, hof_10, log_10), (30, hof_30, log_30), (50, hof_50, log_50)]:
    mejor_fitness = hof[0].fitness.values[0]
    mejor_x = hof[0][0]
    mejor_y = hof[0][1]
    avg_final = log[-1]["avg"]
    std_final = log[-1]["std"]
    error_pct = abs(optimo - mejor_fitness) / optimo * 100
    print(f"{tp:<12} {mejor_fitness:<16.6f} {mejor_x:<14.6f} {mejor_y:<14.6f} {avg_final:<13.4f} {std_final:<12.4f} {error_pct:<10.4f}")
```

## Análisis de los resultados

**Observaciones:**

1. **A mayor tamaño de población, el algoritmo se acerca más al valor óptimo.** La población de 50 individuos logró el menor porcentaje de error respecto al valor óptimo teórico.

2. **El valor promedio (avg) de la población se incrementa a lo largo de las generaciones.** Esto es positivo ya que significa que los individuos que forman la población son, en media, mejores.

3. **La desviación estándar (std) va disminuyendo en general.** Esto indica que los individuos cada vez son más parecidos, lo cual es esperable ya que el algoritmo genético es elitista por el proceso de selección.

4. **El valor máximo (max) va aumentando a lo largo de las generaciones.** Esto indica que el algoritmo está funcionando correctamente, encontrando mejores soluciones progresivamente.

5. **Los picos negativos** en el valor mínimo se deben a que las operaciones genéticas (cruza y mutación) pueden producir individuos fuera del dominio [-100, 100], los cuales reciben un fitness de -1.

6. **Mayor diversidad genética inicial** (poblaciones más grandes) potencia la capacidad de exploración y explotación del algoritmo, permitiendo encontrar soluciones más cercanas al óptimo.
