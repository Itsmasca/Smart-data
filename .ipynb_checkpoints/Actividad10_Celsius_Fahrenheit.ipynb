En esta actividad construiremos una red neuronal con TensorFlow/Keras capaz de **aprender** la conversión de Celsius a Fahrenheit a partir de datos, sin utilizar la fórmula directamente.

## Paso 1 – Importar las librerías

```python
# EM | librerías base: tf para la red, pandas para el csv, sns/plt para visualizar
import tensorflow as tf
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("TensorFlow version:", tf.__version__)
print("Librerías importadas correctamente ✓")
```

## Paso 2 – Obtención de los datos
Cargamos el archivo `celsius_a_fahrenheit.csv` y mostramos los primeros 5 registros.

```python
# EM | cargo el csv con los 30 pares celsius-fahrenheit que usaré para entrenar
df_temperatura = pd.read_csv('celsius_a_fahrenheit.csv')

print("Forma del dataset:", df_temperatura.shape)
print("\nPrimeros 5 datos:")
df_temperatura.head()
```

## Paso 3 – Correlación de los datos

```python
# EM | scatter con seaborn para confirmar que la relación es lineal antes de modelar
plt.figure(figsize=(8, 5))
sns.scatterplot(x='Celsius', y='Fahrenheit', data=df_temperatura, color='royalblue', s=80)
plt.title('Correlación entre Celsius y Fahrenheit', fontsize=14)
plt.xlabel('Celsius')
plt.ylabel('Fahrenheit')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Paso 4 – Selección de variables

```python
# EM | X = entrada (celsius), y = lo que quiero predecir (fahrenheit)
X_train = df_temperatura['Celsius'].values.astype(float)
y_train = df_temperatura['Fahrenheit'].values.astype(float)

print("Variable independiente (Celsius):", X_train[:5])
print("Variable objetivo   (Fahrenheit):", y_train[:5])
```

# 📌 Pregunta 1 – Optimizadores: Adam, SGD y RMSProp

### Adam (Adaptive Moment Estimation)
Es el optimizador utilizado en el modelo base. Combina las ventajas de **AdaGrad** y **RMSProp**: calcula tasas de aprendizaje adaptativas para cada parámetro usando estimaciones del primero y segundo momento del gradiente. Tiende a converger más rápido y es robusto frente a datos ruidosos. En nuestro modelo produce una pérdida que cae rápidamente durante las primeras épocas.

### SGD (Stochastic Gradient Descent)
Actualiza los pesos usando el gradiente calculado sobre un mini-lote (o un solo ejemplo). Es el optimizador más sencillo: `w = w - lr * ∂L/∂w`. Con SGD el entrenamiento puede ser más lento y errático, requiriendo más épocas para converger. Sin embargo, con una tasa de aprendizaje bien afinada puede alcanzar mínimos más generalizables.

### RMSProp (Root Mean Square Propagation)
Mantiene una media móvil del cuadrado del gradiente para normalizar la actualización: `w = w - lr / sqrt(E[g²] + ε) * g`. Esto evita tasas de aprendizaje demasiado grandes o pequeñas. Es especialmente eficaz en problemas de series de tiempo o datos con gradientes muy variables.

### ¿Cómo cambian los resultados?
- **SGD**: converge más lento, necesita más épocas (ej. 500–1000) para llegar a la misma pérdida que Adam en 100 épocas.
- **RMSProp**: convergencia similar a Adam pero sin el momentum del primer momento; puede oscilar más en los últimos pasos.
- **Adam**: mejor balance velocidad-estabilidad para este problema lineal simple.

# 📌 Pregunta 2 – Función de pérdida: mean_squared_error y alternativas

### ¿Por qué es importante `mean_squared_error`?
`MSE = (1/n) Σ(ŷᵢ - yᵢ)²`  
El MSE penaliza errores grandes de forma cuadrática, lo que obliga al modelo a reducir predicciones muy alejadas del valor real. Es diferenciable en todos los puntos, facilitando el cálculo del gradiente. Para un problema de regresión como conversión de temperaturas, es la opción natural.

### Otras funciones de pérdida que se pueden implementar:

| Función | Fórmula simplificada | Efecto en el modelo |
|---|---|---|
| `mean_absolute_error` (MAE) | `(1/n) Σ|ŷᵢ - yᵢ|` | Menos sensible a outliers; gradientes constantes pueden hacer más lenta la convergencia |
| `mean_squared_logarithmic_error` (MSLE) | `(1/n) Σ(log(ŷ+1) - log(y+1))²` | Útil cuando los valores tienen escala muy diferente; penaliza menos errores en valores grandes |
| `huber` | Combina MAE y MSE según un umbral δ | Balance entre robustez ante outliers (MAE) y gradientes suaves (MSE) |

**¿Cómo alteraría los resultados?**  
Con MAE el modelo puede converger a una solución igualmente precisa pero con gradientes más ruidosos, pudiendo necesitar más épocas. Con Huber se obtiene un modelo más robusto si el CSV tuviera valores atípicos. Para este dataset limpio y lineal, MSE sigue siendo la elección óptima.

## Modelo A – 1 Capa (configuración original del PDF)

```python
# EM | modelo base: 1 neurona entrada → 1 salida, suficiente para una relación lineal
tf.random.set_seed(42)

salida_1 = tf.keras.layers.Dense(units=1)
modelo_1capa = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(1,)),
    salida_1
])

# Adam lr=1.0 + MSE: Adam ajusta bien para regresión, MSE penaliza errores grandes
modelo_1capa.compile(
    optimizer=tf.keras.optimizers.Adam(1.0),
    loss='mean_squared_error'
)

modelo_1capa.summary()
```

```python
# EM | verbose=False para no imprimir las 500 épocas, solo me interesa el resultado final
print("Entrenando Modelo 1 Capa...")
hist_1capa = modelo_1capa.fit(X_train, y_train, epochs=500, verbose=False)

loss_final_1capa = hist_1capa.history['loss'][-1]
print(f"Loss final (época 500): {loss_final_1capa:.4f}")

# detecto en qué época la pérdida ya era aceptable (< 1.0)
losses = hist_1capa.history['loss']
epoca_estable = next((i+1 for i, l in enumerate(losses) if l < 1.0), None)
print(f"Pérdida < 1.0 alcanzada en época: {epoca_estable}")
```

---
## Modelo B – 3 Capas (con capas ocultas adicionales)

```python
# EM | agrego 2 capas ocultas para comparar contra el modelo simple
# guardo cada capa en variable para poder ver sus pesos después
tf.random.set_seed(42)

oculta1  = tf.keras.layers.Dense(units=3, input_shape=[1])
oculta2  = tf.keras.layers.Dense(units=3)
salida_3 = tf.keras.layers.Dense(units=1)

modelo_3capas = tf.keras.Sequential([oculta1, oculta2, salida_3])

modelo_3capas.compile(
    optimizer=tf.keras.optimizers.Adam(1.0),
    loss='mean_squared_error'
)

modelo_3capas.summary()
```

```python
# EM | mismas épocas que el modelo anterior para que la comparación sea justa
print("Entrenando Modelo 3 Capas...")
hist_3capas = modelo_3capas.fit(X_train, y_train, epochs=500, verbose=False)

loss_final_3capas = hist_3capas.history['loss'][-1]
print(f"Loss final (época 500): {loss_final_3capas:.4f}")

losses_3 = hist_3capas.history['loss']
epoca_estable_3 = next((i+1 for i, l in enumerate(losses_3) if l < 1.0), None)
print(f"Pérdida < 1.0 alcanzada en época: {epoca_estable_3}")
```

---
## Comparación de pérdidas: Modelo 1 Capa vs 3 Capas

```python
# EM | pongo las dos gráficas juntas para ver de un vistazo cuál converge mejor
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(hist_1capa.history['loss'], color='steelblue', linewidth=1.5)
axes[0].set_title('Progreso de Pérdida – Modelo 1 Capa', fontsize=12)
axes[0].set_xlabel('Época')
axes[0].set_ylabel('Training Loss (MSE)')
axes[0].grid(True, alpha=0.3)
axes[0].legend(['Training Loss'])

axes[1].plot(hist_3capas.history['loss'], color='darkorange', linewidth=1.5)
axes[1].set_title('Progreso de Pérdida – Modelo 3 Capas', fontsize=12)
axes[1].set_xlabel('Época')
axes[1].set_ylabel('Training Loss (MSE)')
axes[1].grid(True, alpha=0.3)
axes[1].legend(['Training Loss'])

plt.tight_layout()
plt.show()

print(f"\nResumen comparativo:")
print(f"  Modelo 1 capa  → Loss final: {loss_final_1capa:.4f} | Época estable (< 1.0): {epoca_estable}")
print(f"  Modelo 3 capas → Loss final: {loss_final_3capas:.4f} | Época estable (< 1.0): {epoca_estable_3}")
```

# 📌 Pregunta 3 – ¿Cuántas épocas son suficientes?

Un modelo se considera **suficientemente entrenado** cuando la función de pérdida deja de decrecer de manera significativa entre épocas consecutivas, es decir, cuando ha **convergido**. Un criterio práctico es observar cuándo la pérdida cae por debajo de un umbral aceptable para el problema.

Para este modelo de conversión Celsius–Fahrenheit con el CSV de 30 datos:
- El **Modelo de 1 capa** típicamente alcanza pérdidas por debajo de 1.0 entre las épocas **50 y 150**. Con 500 épocas el modelo está claramente sobre-entrenado en épocas (la mejora adicional es mínima).
- El **Modelo de 3 capas** converge aún más rápido, alcanzando estabilidad entre las épocas **30 y 80** gracias a la mayor capacidad de representación.

**Conclusión:** Para este problema lineal, con ~**100–200 épocas** el modelo ya está listo para producción. Se puede confirmar observando la gráfica de pérdida: cuando la curva es prácticamente horizontal, ya no hay beneficio en continuar entrenando.

---
# 📌 Pregunta 4 – ¿Por qué son importantes las epochs?

Una **época** (epoch) representa un ciclo completo en el que el modelo procesa **todos los datos de entrenamiento** una vez, calcula el error (pérdida) y actualiza sus pesos mediante el optimizador.

Son importantes porque:

1. **Aprendizaje iterativo**: En cada época el modelo ajusta sus pesos para minimizar la pérdida. Con muy pocas épocas el modelo queda **sub-entrenado** (underfitting) y sus predicciones son imprecisas.
2. **Convergencia**: A medida que aumentan las épocas, la pérdida disminuye hasta estabilizarse en un mínimo. Ese punto marca que el modelo ha aprendido el patrón de los datos.
3. **Control del sobre-entrenamiento**: Demasiadas épocas pueden causar **overfitting**, donde el modelo memoriza los datos de entrenamiento pero generaliza mal. Por eso se monitorea la pérdida en validación.
4. **Reproducibilidad**: Fijar el número de épocas garantiza experimentos comparables entre distintas configuraciones de arquitectura u optimizadores.

En pocas palabras, las épocas son el "tiempo de estudio" del modelo: pocas y aprende poco, demasiadas y memoriza en lugar de generalizar.

# 📌 Pregunta 5 – Función de activación: ¿Cuál usa el modelo? ¿Se puede cambiar a ReLU?

### Función de activación original
Las capas `Dense` de Keras usan **activación lineal** por defecto (`activation='linear'`), es decir, **no aplican ninguna transformación no lineal**: `f(x) = x`. Para una regresión lineal como la conversión de temperaturas, esto es correcto porque la relación entre Celsius y Fahrenheit **es lineal** (`F = (9/5)*C + 32`).

### ¿Qué es ReLU?
ReLU (Rectified Linear Unit): `f(x) = max(0, x)`. Introduce no-linealidad al convertir valores negativos a cero. Es la función de activación más usada en capas ocultas de redes profundas.

### ¿Cómo afecta cambiar a ReLU en este problema?
En la **capa de salida** no debería usarse ReLU porque bloquearía predicciones negativas (temperaturas < 0°F). Sin embargo, aplicarla en **capas ocultas** permite al modelo aprender representaciones más complejas. Para este problema lineal, el impacto será mínimo pero demostrable.

```python
# EM | mismo modelo pero con ReLU en capas ocultas — salida sigue lineal para no bloquear negativos
tf.random.set_seed(42)

oculta1_relu = tf.keras.layers.Dense(units=3, activation='relu', input_shape=[1])
oculta2_relu = tf.keras.layers.Dense(units=3, activation='relu')
salida_relu  = tf.keras.layers.Dense(units=1)

modelo_relu = tf.keras.Sequential([oculta1_relu, oculta2_relu, salida_relu])

modelo_relu.compile(
    optimizer=tf.keras.optimizers.Adam(1.0),
    loss='mean_squared_error'
)

modelo_relu.summary()
```

```python
# EM | entreno con ReLU para comparar — spoiler: va a rendir peor por el dying ReLU con valores negativos
print("Entrenando Modelo con ReLU...")
hist_relu = modelo_relu.fit(X_train, y_train, epochs=500, verbose=False)

loss_final_relu = hist_relu.history['loss'][-1]
print(f"Loss final (época 500): {loss_final_relu:.4f}")

losses_relu = hist_relu.history['loss']
epoca_estable_relu = next((i+1 for i, l in enumerate(losses_relu) if l < 1.0), None)
print(f"Pérdida < 1.0 alcanzada en época: {epoca_estable_relu}")
```

```python
# EM | superpongo ambas curvas para ver el impacto real de ReLU vs activación lineal
plt.figure(figsize=(10, 5))
plt.plot(hist_3capas.history['loss'], label='Lineal (default)', color='darkorange', linewidth=1.5)
plt.plot(hist_relu.history['loss'], label='ReLU (capas ocultas)', color='green', linewidth=1.5, linestyle='--')
plt.title('Comparación de Pérdida: Activación Lineal vs ReLU', fontsize=13)
plt.xlabel('Época')
plt.ylabel('Training Loss (MSE)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print(f"\nResultados comparativos (3 capas):")
print(f"  Activación lineal → Loss: {loss_final_3capas:.4f}")
print(f"  Activación ReLU   → Loss: {loss_final_relu:.4f}")
```

## Predicción con el mejor modelo y verificación con fórmula

```python
# EM | prueba con 100°C, el valor correcto es 212°F — así mido qué tan bien aprendió cada modelo
Temp_C = 100
Temp_C_array = np.array([[Temp_C]])  # array 2D requerido por predict()

pred_1capa  = modelo_1capa.predict(Temp_C_array, verbose=0)[0][0]
pred_3capas = modelo_3capas.predict(Temp_C_array, verbose=0)[0][0]
pred_relu   = modelo_relu.predict(Temp_C_array, verbose=0)[0][0]

Temp_F_real = 9/5 * Temp_C + 32  # valor exacto con la fórmula

print(f"Entrada: {Temp_C}°C")
print(f"{'─'*45}")
print(f"Valor real (fórmula):       {Temp_F_real:.4f}°F")
print(f"Modelo 1 capa (lineal):     {pred_1capa:.4f}°F  | Error: {abs(pred_1capa - Temp_F_real):.4f}")
print(f"Modelo 3 capas (lineal):    {pred_3capas:.4f}°F  | Error: {abs(pred_3capas - Temp_F_real):.4f}")
print(f"Modelo 3 capas (ReLU):      {pred_relu:.4f}°F  | Error: {abs(pred_relu - Temp_F_real):.4f}")
```

```python
# EM | grafico las predicciones sobre un rango continuo para ver si la curva que aprendió coincide con la real
celsius_range = np.linspace(-50, 70, 200)
pred_range_1  = modelo_1capa.predict(celsius_range.reshape(-1,1), verbose=0).flatten()
pred_range_3  = modelo_3capas.predict(celsius_range.reshape(-1,1), verbose=0).flatten()
pred_range_r  = modelo_relu.predict(celsius_range.reshape(-1,1), verbose=0).flatten()
real_range    = 9/5 * celsius_range + 32

plt.figure(figsize=(10, 6))
plt.scatter(X_train, y_train, color='black', zorder=5, label='Datos reales', s=60)
plt.plot(celsius_range, real_range,   color='black',      linestyle=':', linewidth=2,   label='Fórmula exacta')
plt.plot(celsius_range, pred_range_1, color='steelblue',  linestyle='-', linewidth=1.5, label='Modelo 1 capa')
plt.plot(celsius_range, pred_range_3, color='darkorange',  linestyle='-', linewidth=1.5, label='Modelo 3 capas (lineal)')
plt.plot(celsius_range, pred_range_r, color='green',      linestyle='--', linewidth=1.5, label='Modelo 3 capas (ReLU)')
plt.title('Predicciones de los modelos vs Datos reales', fontsize=13)
plt.xlabel('Celsius')
plt.ylabel('Fahrenheit')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Conclusiones

1. **Optimizadores**: Adam es la mejor elección para este problema por su convergencia rápida y estable. SGD requeriría más épocas y RMSProp podría oscilar antes de converger.

2. **Función de pérdida**: `mean_squared_error` es ideal aquí. MAE o Huber serían alternativas válidas aunque con convergencia más lenta o diferente sensibilidad a outliers.

3. **Épocas suficientes**: Entre **100 y 200 épocas** son suficientes para este dataset lineal de 30 muestras. La pérdida converge rápidamente y mejoras adicionales son marginales.

4. **Importancia de las épocas**: Son el mecanismo central de aprendizaje. Sin suficientes épocas el modelo no aprende; con demasiadas puede sobreajustarse.

5. **Función de activación**: El modelo base usa activación **lineal** (correcta para regresión lineal). Cambiar a **ReLU** en capas ocultas produce un resultado significativamente peor para este problema. La razón es el fenómeno **Dying ReLU**: muchos valores de Celsius son negativos (-50°C a 0°C), y ReLU los convierte a cero bloqueando el flujo de gradiente. **Lección clave**: ReLU es excelente para problemas no-lineales con entradas positivas, pero para regresiones con rangos negativos la activación lineal (default) es siempre la elección correcta.
