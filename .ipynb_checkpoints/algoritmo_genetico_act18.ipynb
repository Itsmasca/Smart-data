# Actividad 18 — Algoritmo Genético para Inversión en Bonos
### Benítez Iglésias (2014)

**Problema:** Un cliente tiene un capital limitado y debe decidir cuántos bonos de cada tipo comprar para maximizar su rendimiento, respetando la cantidad máxima disponible por bono y sin exceder el capital.

**Partes:**
- **A.** Ejecutar el código del libro y visualizar los bonos comprados
- **B.** Reimplementar con **DEAP** (Distributed Evolutionary Algorithms in Python)
- **C.** Comparar ambas implementaciones en: función objetivo, fitness, operadores y resultados

## Parte A — Código del libro (Benítez Iglésias, 2014)

```python
import random as rnd
from random import random, randint, sample
from collections import namedtuple
import time

rnd.seed(42)  # Semilla para que el resultado sea reproducible


# Calcula el capital invertido por un individuo
def capitalInvertido(individuo):
    return sum(map(lambda x, y: x * y.precio, individuo, inversiones))


# Calcula el rendimiento producido por un individuo
def rendimiento(individuo):
    return sum(map(lambda x, y: x * y.precio * y.rendim,
                   individuo, inversiones))


# Si un individuo gasta más capital del disponible, se eliminan
# aleatoriamente inversiones hasta que se ajusta al capital
def ajustaCapital(individuo):
    ajustado = individuo[:]
    while capitalInvertido(ajustado) > capital:
        pos = randint(0, len(ajustado) - 1)
        if ajustado[pos] > 0:
            ajustado[pos] -= 1
    return ajustado


def creaIndividuo(inversiones, capital):
    individuo = [0] * len(inversiones)
    while capitalInvertido(individuo) < capital:
        eleccion = randint(0, len(inversiones) - 1)
        individuo[eleccion] += 1
    return ajustaCapital(individuo)


def cruza(poblacion, posiciones):
    L = len(poblacion[0])
    hijo   = poblacion[posiciones[0]][:]
    inicio = randint(0, L - 1)
    fin    = randint(inicio + 1, L)
    hijo[inicio:fin] = poblacion[posiciones[1]][inicio:fin]
    return ajustaCapital(hijo)


def muta(individuo, tasaMutacion):
    mutado = []
    for i in range(len(individuo)):
        if random() > tasaMutacion:
            mutado.append(individuo[i])
        else:
            mutado.append(randint(0, inversiones[i].cantidad))
    return ajustaCapital(mutado)


def evoluciona(poblacion, generaciones):
    poblacion.sort(key=lambda x: rendimiento(x))
    N            = len(poblacion)
    tasaMutacion = 0.01
    reproduccion = [x for x in range(N) for y in range(x + 1)]

    historial = []
    for i in range(generaciones):
        padres = sample(reproduccion, 2)
        while padres[0] == padres[1]:
            padres = sample(reproduccion, 2)
        hijos = [cruza(poblacion, padres) for x in range(N - 1)]
        hijos = [muta(x, tasaMutacion) for x in hijos]
        hijos.append(poblacion[-1])  # elitismo
        poblacion = hijos
        poblacion.sort(key=lambda x: rendimiento(x))
        historial.append(rendimiento(poblacion[-1]))

    return poblacion[-1], historial


Inversion = namedtuple('Inversion', 'precio cantidad rendim')

numInver  = 100
maxPrecio = 1000
maxCant   = 10
maxRend   = 0.2

inversiones = [Inversion(random() * maxPrecio, randint(1, maxCant),
                         random() * maxRend) for i in range(numInver)]

capital      = 50000
individuos   = 20
generaciones = 1000

poblacion = [creaIndividuo(inversiones, capital) for i in range(individuos)]

t0 = time.time()
mejor_libro, historial_libro = evoluciona(poblacion, generaciones)
t_libro = time.time() - t0

rend_libro = rendimiento(mejor_libro)
cap_libro  = capitalInvertido(mejor_libro)

print(f"Capital invertido: {cap_libro:.2f} / {capital}")
print(f"Rendimiento:       {rend_libro:.2f}")
print(f"Tiempo:            {t_libro:.2f}s")
```

### Visualización — ¿cuántos bonos debe comprar el cliente?

```python
import matplotlib.pyplot as plt
import numpy as np

# Solo bonos que el cliente debe comprar
indices_comprados = [i for i, c in enumerate(mejor_libro) if c > 0]
cantidades = [mejor_libro[i] for i in indices_comprados]

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Bonos a comprar
axes[0].bar(range(len(indices_comprados)), cantidades, color='steelblue')
axes[0].set_xticks(range(len(indices_comprados)))
axes[0].set_xticklabels([f"B{i}" for i in indices_comprados], rotation=45, fontsize=7)
axes[0].set_xlabel('Bono')
axes[0].set_ylabel('Cantidad comprada')
axes[0].set_title(f'Bonos seleccionados (total={sum(cantidades)}, tipos={len(cantidades)})')

# Evolución del rendimiento por generación
axes[1].plot(historial_libro, color='green')
axes[1].set_xlabel('Generación')
axes[1].set_ylabel('Rendimiento del mejor individuo')
axes[1].set_title('Convergencia del AG (libro)')
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print(f"Bonos distintos comprados: {len(cantidades)}")
print(f"Total de unidades:         {sum(cantidades)}")
```

## Parte B — Reimplementación con DEAP

Se usan los mismos parámetros (capital, número de inversiones, población, generaciones) para que la comparación sea justa.

```python
from deap import base, creator, tools, algorithms
import random as rnd2

rnd2.seed(42)

# --- Fitness y tipo de individuo ---
# weights=(1.0,) → maximización del rendimiento
if 'FitnessMaxDEAP' in dir(creator):
    del creator.FitnessMaxDEAP
if 'IndividualDEAP' in dir(creator):
    del creator.IndividualDEAP

creator.create("FitnessMaxDEAP", base.Fitness, weights=(1.0,))
creator.create("IndividualDEAP", list, fitness=creator.FitnessMaxDEAP)

toolbox = base.Toolbox()

# Un gen es la cantidad a comprar del bono i (0 a inversiones[i].cantidad)
def gen_bono(idx):
    return rnd2.randint(0, inversiones[idx].cantidad)

def crea_individuo_deap():
    ind = creator.IndividualDEAP([gen_bono(i) for i in range(numInver)])
    # Ajustar capital si excede
    while capitalInvertido(ind) > capital:
        pos = rnd2.randint(0, numInver - 1)
        if ind[pos] > 0:
            ind[pos] -= 1
    return ind

toolbox.register("individual", crea_individuo_deap)
toolbox.register("population", tools.initRepeat, list, toolbox.individual)

# --- Función objetivo (fitness) ---
def eval_rendimiento(individuo):
    if capitalInvertido(individuo) > capital:
        return (-1,)  # penalización por violar la restricción
    return (rendimiento(individuo),)

toolbox.register("evaluate", eval_rendimiento)

# --- Operadores ---
# Cruza de dos puntos (DEAP provee una función lista)
toolbox.register("mate", tools.cxTwoPoint)

# Mutación: cambiar aleatoriamente un gen por un nuevo valor válido
def mut_personal(individuo, indpb):
    for i in range(len(individuo)):
        if rnd2.random() < indpb:
            individuo[i] = rnd2.randint(0, inversiones[i].cantidad)
    # Ajustar capital
    while capitalInvertido(individuo) > capital:
        pos = rnd2.randint(0, numInver - 1)
        if individuo[pos] > 0:
            individuo[pos] -= 1
    return (individuo,)

toolbox.register("mutate", mut_personal, indpb=0.01)

# Selección por torneo
toolbox.register("select", tools.selTournament, tournsize=3)

# --- Ejecutar ---
pob_deap = toolbox.population(n=individuos)
hof = tools.HallOfFame(1)
stats = tools.Statistics(lambda ind: ind.fitness.values)
stats.register("max", np.max)
stats.register("avg", np.mean)

t0 = time.time()
pob_deap, logbook = algorithms.eaSimple(
    pob_deap, toolbox,
    cxpb=0.7, mutpb=0.2, ngen=generaciones,
    stats=stats, halloffame=hof, verbose=False
)
t_deap = time.time() - t0

mejor_deap = hof[0]
rend_deap  = rendimiento(mejor_deap)
cap_deap   = capitalInvertido(mejor_deap)
historial_deap = logbook.select("max")

print(f"Capital invertido: {cap_deap:.2f} / {capital}")
print(f"Rendimiento:       {rend_deap:.2f}")
print(f"Tiempo:            {t_deap:.2f}s")
```

## Parte C — Comparación

| Aspecto | Libro (manual) | DEAP |
|---|---|---|
| **Función objetivo** | `rendimiento(individuo)` = Σ cantidad × precio × rendim | Igual, encapsulada en `eval_rendimiento` y devuelta como tupla |
| **Fitness** | Implícito en el ordenamiento por `sort(key=rendimiento)` | Explícito con `creator.FitnessMaxDEAP` y `weights=(1.0,)` |
| **Selección** | Probabilística basada en ranking — mejor individuo tiene más chance | Torneo (`selTournament`, tamaño=3) |
| **Cruza** | Un punto — segmento aleatorio del segundo padre | Dos puntos (`cxTwoPoint`, operador estándar de DEAP) |
| **Mutación** | Probabilidad 0.01 por gen, nuevo valor aleatorio válido | Igual (`mut_personal` personalizada, `indpb=0.01`) |
| **Elitismo** | Se conserva el mejor en la siguiente generación | `HallOfFame(1)` guarda al mejor histórico |
| **Restricción capital** | `ajustaCapital()` quita bonos hasta cumplir | Penalización (-1) + ajuste post-mutación |

```python
# Comparación numérica y gráfica
print(f"{'':20} {'Libro':>15} {'DEAP':>15}")
print("-" * 50)
print(f"{'Rendimiento':<20} {rend_libro:>15.2f} {rend_deap:>15.2f}")
print(f"{'Capital invertido':<20} {cap_libro:>15.2f} {cap_deap:>15.2f}")
print(f"{'Capital disponible':<20} {capital:>15} {capital:>15}")
print(f"{'Bonos distintos':<20} {sum(1 for c in mejor_libro if c>0):>15} {sum(1 for c in mejor_deap if c>0):>15}")
print(f"{'Total unidades':<20} {sum(mejor_libro):>15} {sum(mejor_deap):>15}")
print(f"{'Tiempo (s)':<20} {t_libro:>15.2f} {t_deap:>15.2f}")

# --- Gráfica comparativa ---
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Convergencia
axes[0].plot(historial_libro, label='Libro (manual)', color='green', linewidth=2)
axes[0].plot(historial_deap, label='DEAP', color='blue', linewidth=2)
axes[0].set_xlabel('Generación')
axes[0].set_ylabel('Rendimiento del mejor individuo')
axes[0].set_title('Convergencia del rendimiento')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Comparación de bonos seleccionados
idx_libro = [i for i, c in enumerate(mejor_libro) if c > 0]
idx_deap  = [i for i, c in enumerate(mejor_deap)  if c > 0]
todos_idx = sorted(set(idx_libro) | set(idx_deap))
width = 0.4
x = np.arange(len(todos_idx))

axes[1].bar(x - width/2, [mejor_libro[i] for i in todos_idx], width, label='Libro', color='green')
axes[1].bar(x + width/2, [mejor_deap[i]  for i in todos_idx], width, label='DEAP', color='blue')
axes[1].set_xticks(x)
axes[1].set_xticklabels([f"B{i}" for i in todos_idx], rotation=45, fontsize=6)
axes[1].set_xlabel('Bono')
axes[1].set_ylabel('Cantidad comprada')
axes[1].set_title('Comparación de bonos seleccionados')
axes[1].legend()

plt.tight_layout()
plt.show()
```
