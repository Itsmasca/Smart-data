# Actividad 20 — El Problema del Viajero (TSP) con Colonia de Hormigas

---

## Descripción general

El **Problema del Viajero de Comercio** (Traveling Salesman Problem, TSP) es uno de los problemas clásicos de optimización combinatoria. El enunciado es simple:

> *Dado un conjunto de ciudades y las distancias entre cada par de ellas, encontrar la ruta más corta que visita cada ciudad exactamente una vez y regresa al punto de partida.*

A pesar de su enunciado sencillo, el TSP es un problema **NP-difícil**: el número de rutas posibles crece factorialmente con el número de ciudades. Para $n$ ciudades existen $\frac{(n-1)!}{2}$ rutas distintas, lo que vuelve imposible resolverlo por fuerza bruta para valores moderadamente grandes de $n$ (por ejemplo, con 20 ciudades ya hay más de $10^{17}$ rutas).

Formalmente, si $d_{ij}$ es la distancia entre las ciudades $i$ y $j$, y $\pi$ es una permutación de las $n$ ciudades, se busca minimizar:

$$L(\pi) = \sum_{i=1}^{n-1} d_{\pi(i), \pi(i+1)} + d_{\pi(n), \pi(1)}$$

---

## Objetivos

1. **Minimizar la distancia total recorrida** por el viajero al completar el circuito.
2. **Encontrar una solución de buena calidad en tiempo razonable**, dado que el problema exacto es intratable para muchas ciudades.
3. **Aplicar el algoritmo de Colonia de Hormigas (ACO — Ant Colony Optimization)** como técnica metaheurística inspirada en el comportamiento de las hormigas reales al buscar alimento.
4. **Analizar el efecto de los parámetros** del algoritmo (feromonas, evaporación, número de hormigas) sobre la calidad de la solución encontrada.

---

## Limitaciones

- **Problema NP-difícil**: no existe un algoritmo exacto de complejidad polinomial conocido para resolver el TSP.
- **Dependencia de parámetros**: los algoritmos metaheurísticos como ACO son sensibles a los valores de $\alpha$ (peso de la feromona), $\beta$ (peso de la visibilidad/heurística), y $\rho$ (tasa de evaporación). Un mal ajuste puede llevar a convergencia prematura en un óptimo local.
- **Óptimo local vs global**: ACO no garantiza encontrar el óptimo global, solo una aproximación aceptable.
- **Costo computacional**: aunque es mucho más eficiente que la fuerza bruta, el tiempo de ejecución aumenta rápidamente con el número de ciudades.
- **Versión simétrica**: se asume distancia euclidiana, es decir $d_{ij} = d_{ji}$. El TSP asimétrico (ATSP) es aún más complejo.

---

## Trabajos previos

El estudio del TSP ha motivado el desarrollo de muchas técnicas a lo largo de los años:

| Enfoque | Descripción |
|---|---|
| **Algoritmos exactos** (Programación dinámica de Held-Karp, Branch & Bound) | Garantizan el óptimo pero solo son viables para pocos nodos (n ≤ 20 aprox.) |
| **Heurísticas clásicas** (Vecino más cercano, 2-opt, Lin-Kernighan) | Soluciones rápidas de calidad razonable |
| **Algoritmos genéticos** (Holland, 1975) | Evolución de poblaciones de rutas usando selección, cruce y mutación |
| **Recocido simulado** (Kirkpatrick, 1983) | Basado en el proceso físico de enfriamiento de metales |
| **Colonia de Hormigas — ACO** (Dorigo, 1992) | Inspirado en el rastro de feromonas de hormigas reales |
| **Optimización por Enjambre de Partículas — PSO** (Kennedy & Eberhart, 1995) | Adaptación de PSO al TSP discreto |

El algoritmo **Ant System** propuesto por Marco Dorigo en su tesis doctoral (1992) marcó el nacimiento de la familia ACO. Posteriormente se desarrollaron variantes más sofisticadas como **Ant Colony System** (Dorigo & Gambardella, 1997) y **MAX-MIN Ant System** (Stützle & Hoos, 2000), todas aplicadas con éxito al TSP y otros problemas combinatorios como el ruteo de vehículos (VRP) y la asignación cuadrática (QAP).

## 1. Matriz de distancias

Se genera una matriz simétrica con distancias aleatorias entre ciudades. Como la distancia de A a B es la misma que de B a A, solo llenamos la mitad inferior y la copiamos a la superior.

```python
import random, sys, math

# Genera una matriz de distancias de nCiudades x nCiudades
def matrizDistancias(nCiud, distanciaMaxima):
    matriz = [[0 for i in range(nCiud)] for j in range(nCiud)]
    for i in range(nCiud):
        for j in range(i):
            matriz[i][j] = distanciaMaxima * random.random()
            matriz[j][i] = matriz[i][j]
    return matriz
```

## 2. Cómo elige la hormiga la siguiente ciudad

Esta es la parte central del algoritmo. La hormiga decide a dónde ir combinando dos factores:

- **Feromona** (α = 1.0): qué tanto han pasado otras hormigas por ese camino
- **Distancia** (β = 0.5): qué tan cerca está la ciudad

Cada ciudad disponible recibe un peso, y la hormiga elige una de forma aleatoria pero proporcional a ese peso. Así combina experiencia colectiva con instinto de "lo más cercano".

```python
# Elige un paso de una hormiga combinando feromonas y distancias,
# descartando las ciudades ya visitadas.
def eligeCiudad(dists, ferom, visitadas):
    listaPesos = []
    disponibles = []
    actual = visitadas[-1]

    alfa = 1.0   # peso de las feromonas
    beta = 0.5   # peso de las distancias

    for i in range(len(dists)):
        if i not in visitadas:
            fer = math.pow((1.0 + ferom[actual][i]), alfa)
            peso = math.pow(1.0 / dists[actual][i], beta) * fer
            disponibles.append(i)
            listaPesos.append(peso)

    # Selección proporcional al peso (ruleta)
    valor = random.random() * sum(listaPesos)
    acumulado = 0.0
    i = -1
    while valor > acumulado:
        i += 1
        acumulado += listaPesos[i]

    return disponibles[i]
```

## 3. Construcción del camino de una hormiga

La hormiga arranca siempre en la ciudad 0 y va eligiendo la siguiente usando la función anterior, hasta visitar todas. Al final regresa al origen (por eso se suma la distancia de la última ciudad a la 0).

Devuelve el camino completo y su longitud total.

```python
# Genera el recorrido completo de una hormiga y calcula su longitud
def eligeCamino(distancias, feromonas):
    camino = [0]       # siempre empieza en la ciudad 0
    longCamino = 0

    while len(camino) < len(distancias):
        ciudad = eligeCiudad(distancias, feromonas, camino)
        longCamino += distancias[camino[-1]][ciudad]
        camino.append(ciudad)

    # Regresa a la ciudad de origen
    longCamino += distancias[camino[-1]][0]
    camino.append(0)

    return (camino, longCamino)
```

## 4. Manejo de las feromonas

Dos cosas pasan con las feromonas en cada iteración:

- **Depósito**: cuando una hormiga encuentra un buen camino, deja feromona en cada tramo que recorrió (una "dosis")
- **Evaporación**: todas las feromonas se multiplican por 0.9 para que los caminos viejos pierdan fuerza (evita quedarse atorado en un óptimo local)

```python
# Actualiza la matriz de feromonas siguiendo el camino recibido
def rastroFeromonas(feromonas, camino, dosis):
    for i in range(len(camino) - 1):
        feromonas[camino[i]][camino[i + 1]] += dosis

# Evapora todas las feromonas (coeficiente de evaporación = 0.1)
def evaporaFeromonas(feromonas):
    for lista in feromonas:
        for i in range(len(lista)):
            lista[i] *= 0.9
```

## 5. Algoritmo principal

Este es el ciclo completo del ACO. En cada iteración:

1. Se lanza una hormiga que construye un camino
2. Si el camino es mejor que el mejor encontrado, se guarda
3. La hormiga deja feromonas proporcionales a la calidad de su ruta (caminos más cortos = más feromona)
4. Todas las feromonas se evaporan un poco

Al final se devuelve el mejor camino visto en toda la simulación.

```python
# Resuelve el TSP con el algoritmo de colonia de hormigas
def hormigas(distancias, iteraciones, distMedia):
    n = len(distancias)
    feromonas = [[0 for i in range(n)] for j in range(n)]

    mejorCamino = []
    longMejorCamino = sys.maxsize

    for iter in range(iteraciones):
        (camino, longCamino) = eligeCamino(distancias, feromonas)

        if longCamino <= longMejorCamino:
            mejorCamino = camino
            longMejorCamino = longCamino

        rastroFeromonas(feromonas, camino, distMedia / longCamino)
        evaporaFeromonas(feromonas)

    return (mejorCamino, longMejorCamino)
```

## 6. Ejecución

Se crea una matriz de 10 ciudades con distancias aleatorias (máximo 10) y se corre el algoritmo durante 1000 iteraciones.

La `distMedia` se usa para escalar la dosis de feromona: caminos más cortos que el promedio dejan más feromona, más largos dejan menos.

```python
numCiudades = 10
distanciaMaxima = 10
ciudades = matrizDistancias(numCiudades, distanciaMaxima)

iteraciones = 1000
distMedia = numCiudades * distanciaMaxima / 2

(camino, longCamino) = hormigas(ciudades, iteraciones, distMedia)
print("Camino:", camino)
print("Longitud del camino:", longCamino)
```
