## Reporte de Lectura

### 1.1 La Naturaleza como Optimizadora

La naturaleza ha desarrollado, a lo largo de millones de años, mecanismos sorprendentemente eficientes para resolver problemas de optimización. El proceso evolutivo que observamos en los seres vivos es, en esencia, una búsqueda continua de soluciones cada vez mejores ante los retos que impone el entorno.

El concepto central es la **selección natural**: los individuos mejor adaptados a su ambiente tienen mayor probabilidad de sobrevivir y reproducirse, transmitiendo sus características favorables a la siguiente generación. Con el tiempo, la población en conjunto mejora su capacidad de adaptación. Esto es precisamente un proceso de optimización: minimizar el "costo" de sobrevivir en un entorno dado.

Los algoritmos genéticos (AG) imitan este proceso para resolver problemas computacionales difíciles. En lugar de explorar exhaustivamente todas las soluciones posibles, los AG mantienen una **población de soluciones candidatas** y las mejoran iterativamente mediante operadores inspirados en la biología: selección, cruce y mutación.

### 1.2 Un Poco de Biología

Para entender los AG es necesario conocer los conceptos biológicos que los inspiran:

| Concepto biológico | Equivalente en AG |
|---|---|
| **Cromosoma** | Representación codificada de una solución (ej. cadena de bits) |
| **Gen** | Parámetro o variable de la solución |
| **Alelo** | Valor que puede tomar un gen |
| **Individuo / Fenotipo** | Solución decodificada y evaluable |
| **Aptitud (fitness)** | Qué tan buena es la solución al problema |
| **Población** | Conjunto de soluciones candidatas en una generación |
| **Selección** | Elegir los mejores individuos para reproducirse |
| **Cruzamiento (crossover)** | Combinar cromosomas de dos padres para generar hijos |
| **Mutación** | Cambio aleatorio en un gen para introducir diversidad |
| **Generación** | Una iteración completa del ciclo evolutivo |

El ciclo de un algoritmo genético básico es:

```
1. Inicializar población aleatoria
2. Evaluar aptitud de cada individuo
3. Seleccionar padres (los más aptos)
4. Cruzar padres → generar hijos
5. Mutar hijos con baja probabilidad
6. Reemplazar población con hijos
7. Repetir desde 2 hasta alcanzar criterio de parada
```

## Clase `Individuo`

````python
import random

class Individuo:
    """
    Representa una solución candidata dentro de un algoritmo genético.
    
    El cromosoma es una lista de bits (0 o 1) que codifica la solución.
    La aptitud mide qué tan buena es esa solución al problema objetivo.
    """

    def __init__(self, longitud: int, cromosoma: list = None):
        self.longitud = longitud
        # Si no se pasa cromosoma, se genera uno aleatorio (individuo nuevo)
        self.cromosoma = cromosoma if cromosoma else [random.randint(0, 1) for _ in range(longitud)]
        self.aptitud = 0.0

    def calcular_aptitud(self, funcion_objetivo):
        """Evalúa el cromosoma con la función objetivo y guarda el resultado."""
        self.aptitud = funcion_objetivo(self.cromosoma)
        return self.aptitud

    def mutar(self, prob_mutacion: float = 0.01):
        """
        Muta aleatoriamente genes del cromosoma.
        Equivale a un error de copia en la replicación del ADN.
        """
        for i in range(self.longitud):
            if random.random() < prob_mutacion:
                self.cromosoma[i] = 1 - self.cromosoma[i]  # flip del bit

    def cruzar(self, otro: 'Individuo') -> 'Individuo':
        """
        Cruzamiento de un punto: combina genes de dos padres
        para producir un hijo. Simula la reproducción sexual.
        """
        punto = random.randint(1, self.longitud - 1)
        cromosoma_hijo = self.cromosoma[:punto] + otro.cromosoma[punto:]
        return Individuo(self.longitud, cromosoma_hijo)

    def __repr__(self):
        bits = ''.join(map(str, self.cromosoma))
        return f"Individuo(cromosoma={bits}, aptitud={self.aptitud:.4f})"
````

## Clase `Poblacion`

````python
class Poblacion:
    """
    Representa el conjunto de individuos (soluciones candidatas)
    que evolucionan a lo largo de generaciones.
    
    Implementa el ciclo evolutivo completo: evaluación, selección,
    cruzamiento y mutación.
    """

    def __init__(self, tamaño: int, longitud_cromosoma: int):
        self.tamaño = tamaño
        self.longitud_cromosoma = longitud_cromosoma
        self.generacion = 0
        # Crear población inicial con individuos aleatorios
        self.individuos = [Individuo(longitud_cromosoma) for _ in range(tamaño)]

    def evaluar(self, funcion_objetivo):
        """Calcula la aptitud de todos los individuos de la población."""
        for individuo in self.individuos:
            individuo.calcular_aptitud(funcion_objetivo)

    def seleccionar(self) -> Individuo:
        """
        Selección por torneo: elige 3 individuos al azar y devuelve el más apto.
        Los más aptos tienen mayor probabilidad de reproducirse (como en la naturaleza).
        """
        competidores = random.sample(self.individuos, min(3, self.tamaño))
        return max(competidores, key=lambda ind: ind.aptitud)

    def evolucionar(self, prob_mutacion: float = 0.01):
        """
        Genera una nueva generación mediante selección, cruzamiento y mutación.
        Reemplaza la población actual con los descendientes.
        """
        nueva_generacion = []
        for _ in range(self.tamaño):
            padre = self.seleccionar()
            madre = self.seleccionar()
            hijo = padre.cruzar(madre)
            hijo.mutar(prob_mutacion)
            nueva_generacion.append(hijo)
        self.individuos = nueva_generacion
        self.generacion += 1

    def mejor_individuo(self) -> Individuo:
        """Devuelve el individuo con mayor aptitud de la generación actual."""
        return max(self.individuos, key=lambda ind: ind.aptitud)

    def aptitud_promedio(self) -> float:
        """Calcula la aptitud promedio de la población."""
        return sum(ind.aptitud for ind in self.individuos) / self.tamaño

    def __repr__(self):
        return (f"Poblacion(tamaño={self.tamaño}, "
                f"generacion={self.generacion}, "
                f"mejor_aptitud={self.mejor_individuo().aptitud:.4f}, "
                f"aptitud_promedio={self.aptitud_promedio():.4f})")
````

## Demostración: Problema OneMax

El **problema OneMax** consiste en encontrar la cadena de bits con todos los valores en 1.  
Es el "Hola Mundo" de los algoritmos genéticos: simple de verificar pero útil para validar que la evolución funciona.

````python
import matplotlib.pyplot as plt
import seaborn as sns

# Función objetivo: contar cuántos 1s tiene el cromosoma (OneMax)
def onemax(cromosoma):
    return sum(cromosoma) / len(cromosoma)  # normalizado entre 0 y 1

# Parámetros del AG
TAMAÑO_POBLACION = 50
LONGITUD_CROMOSOMA = 20
GENERACIONES = 30
PROB_MUTACION = 0.02

# Inicializar población
poblacion = Poblacion(TAMAÑO_POBLACION, LONGITUD_CROMOSOMA)

# Evolucionar y registrar progreso
historial_mejor = []
historial_promedio = []

for gen in range(GENERACIONES):
    poblacion.evaluar(onemax)
    historial_mejor.append(poblacion.mejor_individuo().aptitud)
    historial_promedio.append(poblacion.aptitud_promedio())
    poblacion.evolucionar(PROB_MUTACION)

# Evaluar última generación
poblacion.evaluar(onemax)

print(f"Generaciones ejecutadas: {poblacion.generacion}")
print(f"Mejor individuo final:   {poblacion.mejor_individuo()}")
print(f"Aptitud promedio final:  {poblacion.aptitud_promedio():.4f}")

# Graficar evolución
fig, ax = plt.subplots(figsize=(10, 5))
sns.lineplot(x=range(GENERACIONES), y=historial_mejor, ax=ax, label='Mejor aptitud', color='blue')
sns.lineplot(x=range(GENERACIONES), y=historial_promedio, ax=ax, label='Aptitud promedio', color='orange')
ax.set_xlabel("Generación")
ax.set_ylabel("Aptitud")
ax.set_title("Evolución de la aptitud por generación (OneMax)")
ax.set_ylim([0, 1.05])
plt.legend()
plt.tight_layout()
plt.show()
````
