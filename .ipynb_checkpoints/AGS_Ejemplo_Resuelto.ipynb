# Algoritmo Genético Simple (AGS) - Ejemplo Resuelto

**Maximizar la función:**

$$f(x, y) = 21.5 + x \cdot \sin(4\pi x) + y \cdot \sin(20\pi y)$$

**Características:**
- Variables: x ∈ [-3.0, 12.1], y ∈ [4.1, 5.8]
- Precisión: 4 decimales
- Representación binaria
- Selección proporcional por ruleta
- Cruza de un punto
- Mutación aleatoria

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Funciones del AGS

```python
def creapob(Nind, Lind):
    """Crea una población aleatoria de individuos binarios."""
    return np.random.randint(0, 2, size=(Nind, Lind))
```

```python
def decodifica(genotipo, rango):
    """Decodifica el genotipo binario al fenotipo (valores reales)."""
    Nvar = rango.shape[1]
    Nind, Lind = genotipo.shape
    Lvar = Lind // Nvar
    potencias = 2 ** np.arange(Lvar)

    fenotipo = np.zeros((Nind, Nvar))

    for i in range(Nind):
        for j in range(Nvar):
            fenotipo[i, j] = np.sum(potencias * genotipo[i, (j * Lvar):(j * Lvar + Lvar)])

    for i in range(Nvar):
        fenotipo[:, i] = rango[i, 0] + ((rango[i, 1] - rango[i, 0]) / (2 ** Lvar - 1)) * fenotipo[:, i]

    return fenotipo
```

```python
def objfun(fenotipo):
    """Evalúa la función objetivo f(x,y) = 21.5 + x*sin(4*pi*x) + y*sin(20*pi*y)."""
    Nind, Nvar = fenotipo.shape
    objv = np.zeros((Nind, 1))

    for i in range(Nind):
        objv[i, 0] = (21.5
                       + fenotipo[i, 0] * np.sin(4 * np.pi * fenotipo[i, 0])
                       + fenotipo[i, 1] * np.sin(20 * np.pi * fenotipo[i, 1]))

    return objv
```

```python
def rankeo(objv, direccion):
    """Transforma el valor objetivo a un valor de aptitud mediante rankeo."""
    SP = 2
    Nind, Nobj = objv.shape
    aptitud = np.zeros((Nind, 1))

    if direccion == 1:
        nuevo_objv = np.sort(objv, axis=0)
    else:
        nuevo_objv = np.sort(-1 * objv, axis=0)

    apt = 2 - SP + 2 * (SP - 1) * (np.arange(Nind) / (Nind - 1))
    posori = np.argsort(objv, axis=0).flatten()

    aptitud[posori, 0] = apt.reshape(-1)

    return aptitud
```

```python
def ruleta(genotipo, fenotipo, aptitud):
    """Selección proporcional por ruleta."""
    Nind, aux = aptitud.shape
    total = np.sum(aptitud)
    probabilidad = aptitud / total
    acumulada = np.cumsum(probabilidad)
    idx = np.zeros((Nind, 1), dtype=int)

    for i in range(Nind):
        selecciona = np.random.rand()
        aux = np.where(acumulada >= selecciona)[0]
        idx[i, 0] = aux[0]

    nuevo_gen = genotipo[idx.flatten(), :]

    return nuevo_gen
```

```python
def xunpunto(nuevo_gen, Pc):
    """Cruza de un punto."""
    Nind, Lind = nuevo_gen.shape
    aux_gen = np.zeros((Nind, Lind))
    par = Nind % 2

    for i in range(0, Nind - 1, 2):
        cruza = np.random.rand()
        if cruza <= Pc:
            corte = np.random.randint(0, Lind - 1)
            aux_gen[i, :] = np.concatenate((nuevo_gen[i, :corte], nuevo_gen[i + 1, corte:]))
            aux_gen[i + 1, :] = np.concatenate((nuevo_gen[i + 1, :corte], nuevo_gen[i, corte:]))
        else:
            aux_gen[i, :] = nuevo_gen[i, :]
            aux_gen[i + 1, :] = nuevo_gen[i + 1, :]

    if par == 1:
        aux_gen[Nind - 1, :] = nuevo_gen[Nind - 1, :]

    nuevo_gen = aux_gen

    return nuevo_gen
```

```python
def muta(nuevo_gen, Pm):
    """Mutación aleatoria uniforme."""
    Nind, Lind = nuevo_gen.shape
    valores = np.random.rand(Nind, Lind)
    muta_mask = valores <= Pm
    nuevo_gen = np.logical_xor(nuevo_gen, muta_mask).astype(int)

    return nuevo_gen
```

## 1. Definir los parámetros del AGS

```python
# Valores de las variables
Nind = 100          # Número de individuos en la población
Lind = 33           # Longitud de los individuos (18 bits para x + 15 bits para y)
Pc = 0.9            # Probabilidad de cruzamiento
Pm = 0.01           # Probabilidad de mutación
Maxgen = 1000       # Máximo número de generaciones
Nvar = 2            # Número de variables
rango = np.array([[-3, 4.1], [12.1, 5.8]])  # Rangos de las variables
```

## 2. Crear la población inicial

```python
genotipo = creapob(Nind, Lind)
fenotipo = decodifica(genotipo, rango)
generaciones = 1
```

## 3. Obtener el valor óptimo

```python
Mejor = np.full((Maxgen, 1), np.nan)              # Mejor valor óptimo por generación
Mejor_cromosoma = np.zeros((Maxgen, Lind))         # Mejor cromosoma por generación
objv = objfun(fenotipo)                            # Evaluación de la función
```

## 4. Ciclo Evolutivo

```python
generaciones = 1

while generaciones < Maxgen:
    # Rankeo: transformar valor objetivo a aptitud
    aptitud = rankeo(objv, 1)

    # Selección por ruleta
    nuevo_gen = ruleta(genotipo, fenotipo, aptitud)

    # Cruza de un punto
    nuevo_gen = xunpunto(nuevo_gen, Pc)

    # Mutación
    nuevo_gen = muta(nuevo_gen, Pm)

    # Decodificar y evaluar nueva población
    nuevo_feno = decodifica(nuevo_gen, rango)
    nuevo_objv = objfun(nuevo_feno)

    # Sustituir población anterior
    genotipo = nuevo_gen
    objv = nuevo_objv

    # Almacenar mejor solución de esta generación
    idx = np.argmax(objv)
    Mejor[generaciones] = objv[idx]
    Mejor_cromosoma[generaciones, :] = genotipo[idx, :]

    generaciones += 1
```

## 5. Resultados

```python
mejor_gen = np.nanargmax(Mejor)
mejor_valor = Mejor[mejor_gen, 0]
mejor_crom = Mejor_cromosoma[mejor_gen, :].reshape(1, -1)
mejor_feno = decodifica(mejor_crom, rango)

print("=" * 60)
print("ALGORITMO GENÉTICO SIMPLE - EJEMPLO RESUELTO")
print("f(x, y) = 21.5 + x*sin(4*PI*x) + y*sin(20*PI*y)")
print("=" * 60)
print(f"Mejor generación:  {mejor_gen}")
print(f"Mejor valor f(x,y): {mejor_valor:.6f}")
print(f"x = {mejor_feno[0, 0]:.4f}")
print(f"y = {mejor_feno[0, 1]:.4f}")
print("=" * 60)
```

## 6. Gráficas

```python
# Gráfica de la superficie 3D
fig = plt.figure(figsize=(10, 7))
ax1 = fig.add_subplot(111, projection='3d')
x = np.linspace(-3, 12.1, 200)
y = np.linspace(4.1, 5.8, 200)
X, Y = np.meshgrid(x, y)
Z = 21.5 + X * np.sin(4 * np.pi * X) + Y * np.sin(20 * np.pi * Y)
ax1.plot_surface(X, Y, Z, cmap='viridis', alpha=0.8)
ax1.scatter(mejor_feno[0, 0], mejor_feno[0, 1], mejor_valor,
            color='red', s=100, marker='*', label='Óptimo AGS')
ax1.set_xlabel('X')
ax1.set_ylabel('Y')
ax1.set_zlabel('f(x, y)')
ax1.set_title('Gráfica de f(x, y) = 21.5 + x*sin(4πx) + y*sin(20πy)\ncon Algoritmo Genético')
ax1.legend()
plt.tight_layout()
plt.show()
```

```python
# Gráfica de convergencia del AGS
plt.figure(figsize=(10, 5))
generaciones_validas = np.where(~np.isnan(Mejor))[0]
plt.plot(generaciones_validas, Mejor[generaciones_validas], 'b-', linewidth=0.8)
plt.xlabel('Generación')
plt.ylabel('Mejor valor f(x,y)')
plt.title('Convergencia del Algoritmo Genético')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Comparación con programación estructurada (sin AG)

```python
# Definir la función
def f(x, y):
    return 21.5 + x * np.sin(4 * np.pi * x) + y * np.sin(20 * np.pi * y)

# Generar datos para x y y
x = np.linspace(-3, 12.1, 100)
y = np.linspace(4.1, 5.8, 100)
X, Y = np.meshgrid(x, y)
Z = f(X, Y)

# Crear la gráfica en 3D
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, cmap='viridis')

# Etiquetas y título
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('f(x, y)')
ax.set_title('Gráfica de f(x, y) = 21.5 + x * sin(4πx) + y * sin(20πy)')

# Mostrar la gráfica
plt.show()
```
