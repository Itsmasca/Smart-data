# Notebook 1: Regresión Lineal
---

---
## 1. Importación de librerías

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score

import warnings
warnings.filterwarnings('ignore')
plt.style.use('seaborn-v0_8-whitegrid')
SEED = 42
```

---
## 2. Carga y comprensión del dataset real

```python
df = pd.read_csv('wbes_real.csv')
print(f'Dimensiones: {df.shape}  →  {df.shape[0]} países × {df.shape[1]} variables')
print(f'Cobertura: {df["country"].tolist()[:5]} ... y {df.shape[0]-5} más')
df.head(10)
```

```python
print('\n--- Diccionario de variables ---')
var_desc = {
    'country':                'Nombre del país',
    'country_code':           'Código ISO del país',
    'firm_age_years':         'Edad promedio del establecimiento (años)',
    'employees':              'Número promedio de trabajadores',
    'training_pct':           '% de empresas que ofrecen capacitación formal',
    'exporter_pct':           '% de empresas que exportan directamente (≥10% ventas)',
    'access_finance_pct':     '% de empresas que usan bancos para financiar inversiones',
    'competition_pct':        '% de empresas que compiten contra informales',
    'sales_growth':           'Crecimiento real anual de ventas (%) — VARIABLE OBJETIVO',
    'labor_productivity_growth': 'Crecimiento real anual de productividad laboral (%)',
    'export_sales_pct':       '% de ventas totales exportadas directamente',
    'successful':             '1 si sales_growth > mediana global, 0 en otro caso',
}
for k, v in var_desc.items():
    print(f'  {k:<30} → {v}')
```

```python
print('Estadísticas descriptivas:')
df.describe().round(2)
```

```python
print('Valores nulos por columna:')
print(df.isnull().sum())
print(f'\nTotal de registros limpios: {df.dropna().shape[0]} de {df.shape[0]}')
```

---
## 3. Limpieza de datos

```python
feature_cols = ['employees', 'firm_age_years', 'training_pct',
                'access_finance_pct', 'exporter_pct', 'competition_pct']
target = 'sales_growth'

df_clean = df[feature_cols + [target, 'country']].dropna().copy()
print(f'Dataset limpio: {df_clean.shape[0]} países')
print(f'Países eliminados por nulos: {df.shape[0] - df_clean.shape[0]}')
```

---
## 4. Análisis exploratorio — ¿Qué variables explican mejor las ventas?

```python
corr = df_clean[feature_cols + [target]].corr()
corr_target = corr[target].drop(target).sort_values(ascending=True)

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

sns.heatmap(corr, annot=True, fmt='.2f', cmap='RdYlGn', center=0,
            ax=axes[0], square=True)
axes[0].set_title('Matriz de Correlación\n(Dataset WBES Real)', fontsize=13, fontweight='bold')

colors = ['#d73027' if v < 0 else '#1a9850' for v in corr_target]
axes[1].barh(corr_target.index, corr_target.values, color=colors, alpha=0.85)
axes[1].axvline(0, color='black', linewidth=0.8)
axes[1].set_title('Correlación con Crecimiento de Ventas', fontsize=13, fontweight='bold')
axes[1].set_xlabel('Coeficiente de Pearson')

plt.tight_layout()
plt.savefig('correlaciones_real.png', dpi=150, bbox_inches='tight')
plt.show()

print('\nRanking de variables por correlación con sales_growth:')
print(corr_target.sort_values(ascending=False).to_string())
```

```python
top_vars = corr_target.abs().sort_values(ascending=False).head(4).index.tolist()

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
axes = axes.flatten()

for i, var in enumerate(top_vars):
    axes[i].scatter(df_clean[var], df_clean[target], alpha=0.6, s=40,
                    color='steelblue', edgecolors='white', linewidth=0.5)
    z = np.polyfit(df_clean[var], df_clean[target], 1)
    p = np.poly1d(z)
    x_line = np.linspace(df_clean[var].min(), df_clean[var].max(), 100)
    axes[i].plot(x_line, p(x_line), 'r--', linewidth=2)
    # Anotar algunos países
    for _, row in df_clean.nlargest(3, target).iterrows():
        axes[i].annotate(row['country'], (row[var], row[target]),
                         fontsize=7, ha='left', alpha=0.7)
    axes[i].set_xlabel(var, fontsize=11)
    axes[i].set_ylabel('Crecimiento de Ventas (%)', fontsize=11)
    axes[i].set_title(f'{var}\nvs Crecimiento de Ventas', fontsize=12, fontweight='bold')

plt.suptitle('Relación entre variables predictoras y Ventas\n(Datos Reales WBES)',
             fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('scatter_predictoras_real.png', dpi=150, bbox_inches='tight')
plt.show()
```

---
## 5. Construcción del modelo de Regresión Lineal

```python
X = df_clean[feature_cols]
y = df_clean[target]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=SEED
)
print(f'Train: {X_train.shape[0]} países | Test: {X_test.shape[0]} países')

model = LinearRegression()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
```

---
## 6. Métricas del modelo

```python
r2   = r2_score(y_test, y_pred)
mse  = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)

print('=' * 55)
print('       MÉTRICAS — REGRESIÓN LINEAL (WBES REAL)')
print('=' * 55)
print(f'  R²   (Varianza explicada):   {r2:.4f}  ({r2*100:.1f}%)')
print(f'  MSE  (Error cuadrático):     {mse:.4f}')
print(f'  RMSE (Raíz del MSE):         {rmse:.4f} puntos porcentuales')
print(f'  Intercepto (β₀):             {model.intercept_:.4f}')
print('=' * 55)

coef_df = pd.DataFrame({'Variable': feature_cols, 'Coeficiente': model.coef_})
coef_df = coef_df.sort_values('Coeficiente', ascending=False)
print('\nCoeficientes del modelo:')
print(coef_df.to_string(index=False))
```

---
## 7. Visualizaciones

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

# Real vs Predicho
axes[0].scatter(y_test, y_pred, alpha=0.7, s=60, color='steelblue', edgecolors='white')
lims = [min(y_test.min(), y_pred.min()), max(y_test.max(), y_pred.max())]
axes[0].plot(lims, lims, 'r--', lw=2, label='Predicción perfecta')
axes[0].set_xlabel('Crecimiento real de ventas (%)')
axes[0].set_ylabel('Crecimiento predicho (%)')
axes[0].set_title(f'Real vs Predicho\nR² = {r2:.4f}', fontsize=13, fontweight='bold')
axes[0].legend()

# Residuos
residuals = y_test - y_pred
axes[1].scatter(y_pred, residuals, alpha=0.7, s=60, color='coral', edgecolors='white')
axes[1].axhline(0, color='black', lw=1.5, ls='--')
axes[1].set_xlabel('Valores predichos')
axes[1].set_ylabel('Residuos')
axes[1].set_title('Gráfica de Residuos', fontsize=13, fontweight='bold')

# Coeficientes
colors_coef = ['#1a9850' if c > 0 else '#d73027' for c in coef_df['Coeficiente']]
axes[2].barh(coef_df['Variable'], coef_df['Coeficiente'], color=colors_coef, alpha=0.85)
axes[2].axvline(0, color='black', lw=0.8)
axes[2].set_title('Coeficientes del Modelo', fontsize=13, fontweight='bold')
axes[2].set_xlabel('Valor del coeficiente')

plt.suptitle('Regresión Lineal — Dataset Real WBES 2011–2020',
             fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('regresion_lineal_real.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Top países por crecimiento de ventas
fig, ax = plt.subplots(figsize=(10, 6))
top10 = df_clean.nlargest(10, 'sales_growth')
ax.barh(top10['country'], top10['sales_growth'], color='#1a9850', alpha=0.85)
ax.set_xlabel('Crecimiento real de ventas (%)')
ax.set_title('Top 10 países por Crecimiento de Ventas\n(Promedio 2011–2020)', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.savefig('top10_paises.png', dpi=150, bbox_inches='tight')
plt.show()
```

---
## 8. Interpretación de resultados

### ¿Qué variables explican mejor el crecimiento de ventas?

A partir del análisis de correlación y los coeficientes del modelo sobre datos reales de 113 países:

- training_pct — El porcentaje de empresas que ofrecen capacitación formal es el predictor más consistente. Los países donde más empresas capacitan a su fuerza laboral muestran mayor crecimiento de ventas.
- access_finance_pct — El acceso a financiamiento bancario facilita inversiones en capacidad productiva, lo que se traduce en crecimiento de ventas.
- exporter_pct — Los países con mayor proporción de empresas exportadoras tienden a registrar mayor dinamismo en ventas, dado que acceden a mercados más amplios.
- employees— El tamaño promedio de las empresas refleja economías de escala y mayor capacidad de generación de ingresos.
- competition_pct — La competencia con el sector informal puede presionar los precios y reducir el crecimiento de ventas en algunos contextos.

### Sobre las métricas

El R² obtenido es representativo para un dataset de indicadores agregados por país, donde la variabilidad en el crecimiento de ventas también depende de factores macroeconómicos externos (ciclos económicos, shocks globales como COVID-19 en 2020) que no están capturados en las variables del modelo. El RMSE se interpreta en puntos porcentuales de crecimiento de ventas.

### Nota sobre el dataset

Este análisis utiliza indicadores promedio por país del WBES 2011–2020 (113 economías). Cada observación representa el entorno empresarial promedio de un país, no una empresa individual. Esto le da robustez estadística a las conclusiones para el diseño de políticas públicas comparativas entre países.
