# Optimización por Enjambre de Partículas (PSO)
### Actividad 19

## Función objetivo

La función que se va a minimizar es la **Six-Hump Camel Back**:

$$f(x, y) = x^2 \left(4 - 2.1x^2 + \frac{x^4}{3}\right) + xy + y^2(-4 + 4y^2)$$

Tiene dos mínimos globales en $(0.0898, -0.7126)$ y $(-0.0898, 0.7126)$, ambos con $f \approx -1.0316$

```python
from random import random

def funcion(x, y):
    sum1 = x**2 * (4 - 2.1*x**2 + x**4/3.0)
    sum2 = x*y
    sum3 = y**2 * (-4 + 4*y**2)
    return sum1 + sum2 + sum3

def aleatorio(inf, sup):
    return random()*(sup - inf) + inf
```

## Clase Partícula

Cada partícula tiene posición `(x, y)`, velocidad `(vx, vy)` y su mejor posición encontrada `(xLoc, yLoc)`. Los parámetros `inercia`, `cognitiva` y `social` controlan cómo se mueve.

```python
class Particula:
    inercia   = 1.4
    cognitiva = 2.0
    social    = 2.0
    infx = -2.0
    supx =  2.0
    infy = -1.0
    supy =  1.0
    ajusteV = 100.0

    def __init__(self):
        self.x = aleatorio(Particula.infx, Particula.supx)
        self.y = aleatorio(Particula.infy, Particula.supy)
        self.vx = aleatorio(Particula.infx/Particula.ajusteV,
                            Particula.supx/Particula.ajusteV)
        self.vy = aleatorio(Particula.infy/Particula.ajusteV,
                            Particula.supy/Particula.ajusteV)
        self.xLoc     = self.x
        self.yLoc     = self.y
        self.valorLoc = funcion(self.x, self.y)

    def actualizaVelocidad(self, xGlob, yGlob):
        cogX = Particula.cognitiva * random() * (self.xLoc - self.x)
        socX = Particula.social    * random() * (xGlob - self.x)
        self.vx = Particula.inercia * self.vx + cogX + socX
        cogY = Particula.cognitiva * random() * (self.yLoc - self.y)
        socY = Particula.social    * random() * (yGlob - self.x)
        self.vy = Particula.inercia * self.vy + cogY + socY

    def actualizaPosicion(self):
        self.x = self.x + self.vx
        self.y = self.y + self.vy
        self.x = max(self.x, Particula.infx)
        self.x = min(self.x, Particula.supx)
        self.y = max(self.y, Particula.infy)
        self.y = min(self.y, Particula.supy)
        valor = funcion(self.x, self.y)
        if valor < self.valorLoc:
            self.xLoc     = self.x
            self.yLoc     = self.y
            self.valorLoc = valor
```

## Algoritmo principal

```python
def enjambreParticulas(particulas, iteraciones, reduccionInercial):
    mejorParticula = min(particulas, key=lambda p: p.valorLoc)
    xGlob    = mejorParticula.xLoc
    yGlob    = mejorParticula.yLoc
    valorGlob = mejorParticula.valorLoc

    for iter in range(iteraciones):
        for p in particulas:
            p.actualizaVelocidad(xGlob, yGlob)
            p.actualizaPosicion()

        # Se actualiza el mínimo global solo después de mover todas las partículas
        mejorParticula = min(particulas, key=lambda p: p.valorLoc)
        if mejorParticula.valorLoc < valorGlob:
            xGlob     = mejorParticula.xLoc
            yGlob     = mejorParticula.yLoc
            valorGlob = mejorParticula.valorLoc

        Particula.inercia *= reduccionInercial

    return (xGlob, yGlob, valorGlob)
```

## Ejecución

```python
nParticulas = 10
iteraciones = 100
redInercia  = 0.9

particulas = [Particula() for i in range(nParticulas)]
resultado = enjambreParticulas(particulas, iteraciones, redInercia)

print(f"x       = {resultado[0]:.6f}")
print(f"y       = {resultado[1]:.6f}")
print(f"f(x, y) = {resultado[2]:.6f}")
print(f"Error   = {abs(resultado[2] - (-1.0316)):.6f}")
```

## Visualización

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm

x_vals = np.linspace(-2, 2, 300)
y_vals = np.linspace(-1, 1, 300)
X, Y = np.meshgrid(x_vals, y_vals)
Z = X**2 * (4 - 2.1*X**2 + X**4/3.0) + X*Y + Y**2 * (-4 + 4*Y**2)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

ax1 = axes[0]
cp = ax1.contourf(X, Y, Z, levels=50, cmap='viridis')
fig.colorbar(cp, ax=ax1)
ax1.contour(X, Y, Z, levels=20, colors='white', linewidths=0.4, alpha=0.5)
ax1.scatter(resultado[0], resultado[1], color='red', s=120,
            zorder=5, label=f'PSO: ({resultado[0]:.3f}, {resultado[1]:.3f})')
ax1.scatter([0.0898, -0.0898], [-0.7126, 0.7126], color='lime', s=80,
            marker='*', zorder=5, label='Mínimos globales reales')
ax1.set_title('Six-Hump Camel Back — Vista superior')
ax1.set_xlabel('x')
ax1.set_ylabel('y')
ax1.legend(fontsize=8)

ax2 = fig.add_subplot(122, projection='3d')
ax2.plot_surface(X, Y, Z, cmap='viridis', alpha=0.8, linewidth=0)
ax2.scatter(resultado[0], resultado[1], resultado[2], color='red', s=80)
ax2.set_title('Six-Hump Camel Back — Vista 3D')
ax2.set_xlabel('x')
ax2.set_ylabel('y')
ax2.set_zlabel('f(x,y)')

plt.tight_layout()
plt.show()
```

## Análisis de convergencia

Se ejecuta el PSO 30 veces para ver qué tan seguido encuentra el mínimo global.

```python
N_ejecuciones = 30
resultados = []

for _ in range(N_ejecuciones):
    Particula.inercia = 1.4
    pts = [Particula() for _ in range(nParticulas)]
    res = enjambreParticulas(pts, iteraciones, redInercia)
    resultados.append(res[2])

resultados = np.array(resultados)
minimo_global = -1.0316
tolerancia = 0.05
exitos = np.sum(np.abs(resultados - minimo_global) < tolerancia)

print(f"Mejor:  {resultados.min():.6f}")
print(f"Peor:   {resultados.max():.6f}")
print(f"Media:  {resultados.mean():.6f}")
print(f"Std:    {resultados.std():.6f}")
print(f"Éxitos: {exitos}/{N_ejecuciones}")

plt.figure(figsize=(8, 4))
plt.hist(resultados, bins=15, color='steelblue', edgecolor='white', alpha=0.8)
plt.axvline(minimo_global, color='red', linestyle='--', linewidth=2, label=f'Mínimo global ({minimo_global})')
plt.xlabel('f(x, y)')
plt.ylabel('Frecuencia')
plt.title(f'Resultados PSO — {N_ejecuciones} ejecuciones')
plt.legend()
plt.tight_layout()
plt.show()
```

## Efecto del tamaño del enjambre

```python
tamanos = [5, 10, 20, 50, 100]
medias  = []
stds    = []

for n in tamanos:
    vals = []
    for _ in range(20):
        Particula.inercia = 1.4
        pts = [Particula() for _ in range(n)]
        res = enjambreParticulas(pts, iteraciones, redInercia)
        vals.append(res[2])
    medias.append(np.mean(vals))
    stds.append(np.std(vals))

plt.figure(figsize=(8, 4))
plt.errorbar(tamanos, medias, yerr=stds, fmt='o-', capsize=5,
             color='steelblue', ecolor='gray', linewidth=2, markersize=8)
plt.axhline(minimo_global, color='red', linestyle='--', label='Mínimo global')
plt.xlabel('Número de partículas')
plt.ylabel('f(x, y) promedio ± std')
plt.title('Efecto del tamaño del enjambre')
plt.legend()
plt.tight_layout()
plt.show()

for n, m, s in zip(tamanos, medias, stds):
    print(f"n={n:3d} → media={m:.4f}, std={s:.4f}")
```

## Conclusiones

- Con más partículas el algoritmo encuentra soluciones más cercanas al óptimo global
- La inercia decreciente hace que al inicio se explore más y al final se refine la solución
- Los valores negativos en el mínimo se dan cuando la mutación o el cruce saca a una partícula del dominio, devolviendo fitness = -1
- PSO es un algoritmo sencillo de implementar y funciona bien en funciones continuas con múltiples mínimos locales

## Solución con Programación Estructurada

Se recorre el espacio de búsqueda con una malla uniforme evaluando la función en cada punto. Es el enfoque más directo: sin heurística ni aleatoriedad.

```python
import time

def busqueda_exhaustiva(pasos):
    infx, supx = -2.0, 2.0
    infy, supy = -1.0, 1.0

    mejor_x    = infx
    mejor_y    = infy
    mejor_val  = funcion(infx, infy)
    evaluaciones = 0

    paso_x = (supx - infx) / pasos
    paso_y = (supy - infy) / pasos

    x = infx
    while x <= supx:
        y = infy
        while y <= supy:
            val = funcion(x, y)
            evaluaciones += 1
            if val < mejor_val:
                mejor_val = val
                mejor_x   = x
                mejor_y   = y
            y += paso_y
        x += paso_x

    return mejor_x, mejor_y, mejor_val, evaluaciones

t0 = time.time()
rx, ry, rval, evals = busqueda_exhaustiva(pasos=1000)
t_est = time.time() - t0

print(f"x       = {rx:.6f}")
print(f"y       = {ry:.6f}")
print(f"f(x, y) = {rval:.6f}")
print(f"Error   = {abs(rval - (-1.0316)):.6f}")
print(f"Evaluaciones: {evals:,}")
print(f"Tiempo: {t_est:.3f}s")
```

## Comparación PSO vs Programación Estructurada

```python
import time

# Ejecutar PSO y medir tiempo
Particula.inercia = 1.4
t0 = time.time()
pts = [Particula() for _ in range(nParticulas)]
res_pso = enjambreParticulas(pts, iteraciones, redInercia)
t_pso = time.time() - t0
evals_pso = nParticulas * iteraciones  # aprox.

# Ejecutar búsqueda exhaustiva y medir tiempo
t0 = time.time()
rx, ry, rval, evals_est = busqueda_exhaustiva(pasos=1000)
t_est = time.time() - t0

minimo_global = -1.0316

print(f"{'':25} {'PSO':>15} {'Exhaustiva':>15}")
print("-" * 55)
print(f"{'x encontrado':<25} {res_pso[0]:>15.6f} {rx:>15.6f}")
print(f"{'y encontrado':<25} {res_pso[1]:>15.6f} {ry:>15.6f}")
print(f"{'f(x,y)':<25} {res_pso[2]:>15.6f} {rval:>15.6f}")
print(f"{'Error absoluto':<25} {abs(res_pso[2]-minimo_global):>15.6f} {abs(rval-minimo_global):>15.6f}")
print(f"{'Evaluaciones aprox.':<25} {evals_pso:>15,} {evals_est:>15,}")
print(f"{'Tiempo (s)':<25} {t_pso:>15.4f} {t_est:>15.4f}")

# Gráfica comparativa
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

for ax, titulo, px, py, pval in [
    (axes[0], 'PSO', res_pso[0], res_pso[1], res_pso[2]),
    (axes[1], 'Programación Estructurada', rx, ry, rval)
]:
    cp = ax.contourf(X, Y, Z, levels=50, cmap='viridis')
    ax.contour(X, Y, Z, levels=20, colors='white', linewidths=0.3, alpha=0.4)
    ax.scatter(px, py, color='red', s=150, zorder=5, label=f'Encontrado: ({px:.3f}, {py:.3f})\nf={pval:.4f}')
    ax.scatter([0.0898, -0.0898], [-0.7126, 0.7126], color='lime', s=80, marker='*', zorder=5, label='Óptimo real')
    ax.set_title(titulo)
    ax.set_xlabel('x')
    ax.set_ylabel('y')
    ax.legend(fontsize=8)

plt.tight_layout()
plt.show()
```
