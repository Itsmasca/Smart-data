```python
import tensorflow as tf 
import tensorflow_datasets as tfds
import matplotlib.pyplot as plt
import seaborn as sns
import math
import numpy as np
```

```python
datos, metadatos = tfds.load('fashion_mnist', as_supervised=True, with_info=True)
```

```python
metadatos
```

```python
datos_entrenamiento, datos_prueba = datos['train'], datos['test']
```

```python
nombres_clases = metadatos.features['label'].names
nombres_clases
```

```python
def normalizar(imagenes, etiqueta):
    imagenes = tf.cast(imagenes, tf.float32)
    imagenes /= 255
    return imagenes, etiqueta
```

```python
datos_entrenamiento = datos_entrenamiento.map(normalizar)
datos_prueba = datos_prueba.map(normalizar)
```

```python
datos_entrenamiento = datos_entrenamiento.cache()
datos_prueba = datos_prueba.cache()
```

```python
# Mostrar una imagen de los datos de entrenamiento
for imagen, etiqueta in datos_entrenamiento.take(1):
    break
imagen = imagen.numpy().reshape((28, 28))

fig, ax = plt.subplots(figsize=(5, 5))
sns.heatmap(imagen, ax=ax, cmap='gray', cbar=True, xticklabels=False, yticklabels=False)
ax.set_title(nombres_clases[etiqueta.numpy()])
plt.show()
```

```python
# Mostrar una imagen por cada categoría con su etiqueta
fig, ejes = plt.subplots(5, 5, figsize=(10, 10))
for i, (imagen, etiqueta) in enumerate(datos_entrenamiento.take(25)):
    imagen = imagen.numpy().reshape((28, 28))
    ax = ejes.flat[i]
    sns.heatmap(imagen, ax=ax, cmap='gray', cbar=False, xticklabels=False, yticklabels=False)
    ax.set_xlabel(nombres_clases[etiqueta.numpy()], fontsize=8)
plt.tight_layout()
plt.show()
```

```python
#Crear el modelo
modelo = tf.keras.Sequential([
    tf.keras.layers.Flatten(input_shape=(28,28,1)), #1 - blanco y negro
    tf.keras.layers.Dense(50, activation=tf.nn.relu),
    tf.keras.layers.Dense(50, activation=tf.nn.relu),
    tf.keras.layers.Dense(10, activation=tf.nn.softmax) #Para redes de clasificacion
])
```

```python
#Compilar el modelo
modelo.compile(
    optimizer='adam',
    loss=tf.keras.losses.SparseCategoricalCrossentropy(),
    metrics=['accuracy']
)
```

```python
TAMANO_LOTE = 32

#Los numeros de datos en entrenamiento y pruebas (60k y 10k)
num_ej_entrenamiento = metadatos.splits["train"].num_examples
num_ej_pruebas = metadatos.splits["test"].num_examples
print("Ejemplos de entrenamiento:", num_ej_entrenamiento)
print("Ejemplos de prueba:", num_ej_pruebas)

#Ajustar propiedades: repeat, shuffle y batch
datos_entrenamiento = datos_entrenamiento.repeat().shuffle(num_ej_entrenamiento).batch(TAMANO_LOTE)
datos_prueba = datos_prueba.batch(TAMANO_LOTE)
```

```python
#Entrenando el modelo
historial = modelo.fit(datos_entrenamiento, epochs=5, steps_per_epoch=math.ceil(num_ej_entrenamiento/TAMANO_LOTE))
```

```python
# Ver la función de pérdida
fig, ax = plt.subplots()
sns.lineplot(data=historial.history["loss"], ax=ax)
ax.set_xlabel("# Época")
ax.set_ylabel("Magnitud de pérdida")
plt.show()
```

```python
# Crear una cuadrícula con varias predicciones, marcar correcta (azul) o incorrecta (roja)
for imagenes_prueba, etiquetas_prueba in datos_prueba.take(1):
    imagenes_prueba = imagenes_prueba.numpy()
    etiquetas_prueba = etiquetas_prueba.numpy()

predicciones = modelo.predict(imagenes_prueba)

def graficar_imagen(i, arr_predicciones, etiquetas_reales, imagenes, ax):
    arr_predicciones, etiqueta_real, img = arr_predicciones[i], etiquetas_reales[i], imagenes[i]
    sns.heatmap(img[..., 0], ax=ax, cmap='gray', cbar=False, xticklabels=False, yticklabels=False)
    etiqueta_prediccion = np.argmax(arr_predicciones)
    color = 'blue' if etiqueta_prediccion == etiqueta_real else 'red'
    ax.set_xlabel("{} {:2.0f}% ({})".format(
        nombres_clases[etiqueta_prediccion],
        100 * np.max(arr_predicciones),
        nombres_clases[etiqueta_real]), color=color, fontsize=7)

def graficar_valor_arreglo(i, arr_predicciones, etiqueta_real, ax):
    arr_predicciones, etiqueta_real = arr_predicciones[i], etiqueta_real[i]
    etiqueta_prediccion = np.argmax(arr_predicciones)
    colores = ['#777777'] * 10
    colores[etiqueta_prediccion] = 'red'
    colores[etiqueta_real] = 'blue'
    sns.barplot(x=list(range(10)), y=arr_predicciones, palette=colores, ax=ax)
    ax.set_ylim([0, 1])
    ax.set_xticks([])
    ax.set_yticks([])

filas = 5
columnas = 5
num_imagenes = filas * columnas
fig, axes = plt.subplots(filas, 2 * columnas, figsize=(2 * 2 * columnas, 2 * filas))
for i in range(num_imagenes):
    graficar_imagen(i, predicciones, etiquetas_prueba, imagenes_prueba, axes.flat[2 * i])
    graficar_valor_arreglo(i, predicciones, etiquetas_prueba, axes.flat[2 * i + 1])
plt.tight_layout()
plt.show()
```

```python
#Realizar la prediccion de una imagen
imagen = imagenes_prueba[24]
imagen = np.array([imagen])

prediccion = modelo.predict(imagen)

print("Prediccion: " + nombres_clases[np.argmax(prediccion[0])])
```

## Actividades

### Actividad 1. Respuestas a las preguntas de la guía

**as_supervised:** Cuando se establece en `True`, los datos se devuelven como tuplas `(imagen, etiqueta)` en lugar de un diccionario. Esto facilita iterar sobre los datos directamente para entrenamiento.

**with_info:** Cuando se establece en `True`, la función `tfds.load()` también devuelve un objeto con metadatos del dataset (número de ejemplos, nombres de clases, forma de las imágenes, etc.).

**map():** Es una función de los datasets de TensorFlow que aplica una transformación (en este caso la función `normalizar`) a cada elemento del dataset. Se usa para que todas las imágenes queden con valores entre 0 y 1.

### Actividad 3. ¿Qué es el formato h5 y para qué sirve?

El formato **HDF5** (Hierarchical Data Format version 5, extensión `.h5`) es un formato de archivo diseñado para almacenar y organizar grandes cantidades de datos numéricos de forma eficiente y jerárquica.

En el contexto de TensorFlow/Keras, el formato h5 se utiliza para **guardar modelos completos** de redes neuronales. Cuando exportas un modelo a `.h5`, se almacena:
- La **arquitectura** del modelo (capas, neuronas, funciones de activación)
- Los **pesos** aprendidos durante el entrenamiento
- El **estado del optimizador**
- La **configuración de compilación** (loss, métricas)

Esto permite reutilizar el modelo entrenado sin tener que volver a entrenarlo, lo cual es muy útil para desplegar modelos en producción.

### Actividad 2. Clasificador interactivo - Cargar imagen y clasificarla

```python
from PIL import Image
import ipywidgets as widgets
from IPython.display import display, clear_output

upload_widget = widgets.FileUpload(accept='.png,.jpg,.jpeg,.bmp', multiple=False, description='Cargar imagen')
output_widget = widgets.Output()

def clasificar_imagen(change):
    with output_widget:
        clear_output(wait=True)

        uploaded_file = list(upload_widget.value.values())[0] if isinstance(upload_widget.value, dict) else upload_widget.value[0]
        contenido = uploaded_file['content'] if isinstance(uploaded_file, dict) else uploaded_file.content

        from io import BytesIO
        img = Image.open(BytesIO(contenido))
        img = img.convert('L')
        img = img.resize((28, 28))

        img_array = np.array(img) / 255.0
        img_array = img_array.reshape(1, 28, 28, 1)

        prediccion = modelo.predict(img_array)
        clase_predicha = np.argmax(prediccion[0])
        confianza = np.max(prediccion[0]) * 100

        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

        sns.heatmap(img_array.reshape(28, 28), ax=ax1, cmap='gray', cbar=False,
                    xticklabels=False, yticklabels=False)
        ax1.set_title("Imagen cargada")

        colores = ['#777777'] * 10
        colores[clase_predicha] = 'blue'
        sns.barplot(x=list(range(10)), y=prediccion[0], palette=colores, ax=ax2)
        ax2.set_xticks(range(10))
        ax2.set_xticklabels(nombres_clases, rotation=45, ha='right', fontsize=8)
        ax2.set_ylim([0, 1])
        ax2.set_title(f"Predicción: {nombres_clases[clase_predicha]} ({confianza:.1f}%)")

        plt.tight_layout()
        plt.show()
        print(f"\nResultado: {nombres_clases[clase_predicha]} con {confianza:.1f}% de confianza")

upload_widget.observe(clasificar_imagen, names='value')

print("Sube una imagen de una prenda de ropa (28x28 px recomendado, escala de grises):")
display(upload_widget)
display(output_widget)
```

```python

modelo.save('modelo_clasificador_ropa.h5')
print("Modelo exportado exitosamente como 'modelo_clasificador_ropa.h5'")

```
